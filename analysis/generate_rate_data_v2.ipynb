```python
# bring back the .mat files from low_perf folder to original folder if needed
low_perf_dir = os.path.join(model_dir, 'low_perf')
if os.path.exists(low_perf_dir):
    folder_names = [f for f in os.listdir(model_dir) if os.path.isdir(os.path.join(model_dir, f)) and f.startswith('Task_')]
    for model_name in folder_names:
        model_path = os.path.join(low_perf_dir, f"{model_name}.mat")
        if os.path.exists(model_path):
            new_path = os.path.join(model_dir, f"{model_name}.mat")
            print(f"Moving {model_name} back to original folder")
            os.rename(model_path, new_path)
```
