```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
T = 300
time = np.arange(0, T)
fs = 200
f = 4
DeltaT = 1

delay_on = 100
delay_off = 150

wave = np.sin(2 * np.pi * f / fs * np.arange(0, delay_off-delay_on) * DeltaT)

plt.plot(wave)
```

```python
T = 300
time = np.arange(0, T)
fs = 200
target_freq = 4
DeltaT = 1

delay_on = 94
delay_off = 144

wave = np.sin(2 * np.pi * f / fs * np.arange(0, delay_off-delay_on) * DeltaT)

lfp_input = np.zeros((1, T), dtype=np.float32)
lfp_input[0, delay_on:delay_off] = wave

plt.plot(lfp_input.T)

wave = np.sin(2 * np.pi * target_freq / fs * time[delay_on:delay_off] * DeltaT)
lfp_input = np.zeros((1, T), dtype=np.float32)
lfp_input[0, delay_on:delay_off] = wave

plt.plot(lfp_input.T)
```
